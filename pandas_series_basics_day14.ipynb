```python
import pandas as pd
```

## Creating Pandas Series

```python
# 1. From a list - default integer index
s1 = pd.Series([10, 20, 30, 40, 50])
print(s1)
```

```python
# 2. With a custom index
marks = pd.Series([85, 90, 78, 92], index=["Asha", "Raj", "Meena", "Vikram"])
print(marks)
```

```python
# 3. From a dictionary - keys become the index
prices = pd.Series({"apple": 40, "banana": 20, "mango": 80})
print(prices)
```

```python
# 4. With a name
scores = pd.Series([1, 2, 3], name="Scores")
print(scores)
```

## Indexing

```python
# by position
print(marks.iloc[0])
print(marks.iloc[-1])
```

```python
# by label
print(marks["Raj"])
print(marks.loc["Meena"])
```

```python
# slicing by position
print(marks[1:3])
```

## Filtering

```python
print(marks[marks > 80])
```

```python
print(marks[marks >= 90])
```

## Arithmetic operations

```python
print(s1 + 10)
```

```python
print(s1 * 2)
```

```python
# arithmetic between two Series aligns by index
bonus = pd.Series([5, 5, 5, 5], index=["Asha", "Raj", "Meena", "Vikram"])
print(marks + bonus)
```

## Basic statistical analysis of a Series

```python
print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Median:", marks.median())
print("Min:", marks.min(), "| Max:", marks.max())
print("Std:", marks.std())
```

```python
print(marks.describe())
```

```python
print(marks.sort_values())
```

```python
print(marks.sort_values(ascending=False))
```

## Interview questions

**What is a Pandas Series?**
A one-dimensional labeled array that can hold any data type. It's built on top of a NumPy array but adds an explicit index, so every value has both a label and a position, and operations align automatically by that label.

**How is a Series different from a DataFrame?**
A Series is a single column of data with one index. A DataFrame is a two-dimensional table made of multiple Series sharing the same index, essentially several columns side by side. Selecting one column from a DataFrame (`df["col"]`) returns a Series.

**How can you filter a Series?**
With a boolean condition inside square brackets: `marks[marks > 80]` returns only the entries where the condition is True, keeping their original labels. The condition itself, `marks > 80`, is a Series of True/False values used as a mask.
